```python
import glob, os, subprocess, sys

root = "/kaggle/input"
hits = sorted(glob.glob(root + "/**/lora_data.json*", recursive=True))
if not hits:
    raise SystemExit(f"dataset not attached. {root} holds: "
                     f"{sorted(os.listdir(root)) if os.path.isdir(root) else 'nothing'}")
data = hits[0]
base = os.path.dirname(data)
print("using", data)
print("alongside it:", sorted(os.listdir(base)))

os.environ["REACTIONGRAPH_DATA"] = data
os.environ["REACTIONGRAPH_OUT"] = "/kaggle/working/outputs"
# base, few_shot and retrieval already finished in the previous run on the same test set,
# same model and same grader; this run does only the stage that failed.
os.environ["REACTIONGRAPH_METHODS"] = "lora"

subprocess.run([sys.executable, os.path.join(base, "run_lora.py")], check=True)
```
